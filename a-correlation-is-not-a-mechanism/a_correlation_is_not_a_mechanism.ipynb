# A Correlation Is Not a Mechanism

Executable companion to Article #35. These nine studies are fully synthetic. The notebook imports the same source functions as `analysis.py` and never changes canonical archived evidence.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from analysis import STUDY_IDS, DEFAULT_SEED, simulate, summarize
from figures import plot_study
print(f'Nine deterministic studies; seed={DEFAULT_SEED}')
```

```python
all_summaries = {}
for study in STUDY_IDS:
    all_summaries[study] = summarize(simulate(study, repetitions=300, seed=DEFAULT_SEED))
    print(f'\n{study}')
    for row in all_summaries[study]:
        print(f"  {row['scenario']:25s} {row['method']:21s} mean={row['mean_estimate']:+.4f} target={row['target']:+.4f} MCSE={row['monte_carlo_se']:.4f}")
```

## Fresh outputs in a temporary directory

Reproduce CSV archives and regenerate all figures without writing into the tracked repository. This cell uses the same command-line entry points as the CI job.

```python
import subprocess
import sys
with TemporaryDirectory(prefix='causal-identification-') as working:
    work = Path(working)
    data_dir, figure_dir = work / 'data', work / 'figures'
    subprocess.run([sys.executable, 'analysis.py', '--output-dir', str(data_dir)], check=True)
    for study in STUDY_IDS:
        plot_study(study, data_dir / f'{study}_summary.csv', figure_dir / f'{study}.png')
    csv_files = list(data_dir.glob('*.csv'))
    png_files = list(figure_dir.glob('*.png'))
    assert len(csv_files) == 18 and len(png_files) == 9
    print(f'Regenerated {len(csv_files)} tables and {len(png_files)} figures without modifying tracked files.')
```

## Interpretation

The displayed Monte Carlo standard errors measure precision of the simulation means. They are not sampling uncertainty estimates for any real study. In the observational-equivalence experiment, two identical bivariate observational laws imply different intervention effects, so no estimator based solely on those observed pairs can identify the ATE.

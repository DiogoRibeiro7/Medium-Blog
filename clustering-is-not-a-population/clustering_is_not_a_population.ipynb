# A Cluster Is Not a Population

Article #28 companion. Nine seeded synthetic studies, ten figures, and eighteen CSV evidence tables. These are conditional comparisons, not estimates of natural or causal population types.


## What the algorithms optimize

- **K-means:** minimizes within-cluster squared Euclidean distance.
- **DBSCAN:** groups points connected by sufficiently dense neighborhoods.
- **Gaussian mixtures:** maximize a likelihood under distributional and covariance assumptions; BIC penalizes model dimension.
- **Silhouette/ARI/stability:** evaluate different targets, none of which automatically identifies scientific populations.

## Reproduce without overwriting canonical evidence

The quick design uses three independent outer replications for every scenario. A temporary output directory isolates its figures and CSVs. Full canonical evidence is regenerated by running the script without the quick flag.


```python
with tempfile.TemporaryDirectory() as directory:
    path = Path(directory)
    summary = run(Design(root=path, quick=True))
    print('Tables:', len(list((path / 'results').glob('*.csv'))))
    print('Figures:', len(list((path / 'figures').glob('*.png'))))
    for figure in FIGURES:
        display(Image(filename=str(path / 'figures' / f'{figure}_quick.png')))
```

## Interpretation

The generator labels supply an **oracle** that is unavailable in ordinary clustering tasks. A good ARI to those labels is not an external causal validation; poor ARI can also reflect an unsuitable target. Bootstrap stability is measured on the same untouched anchor points.

# The Price of Stability Was Not Shared Equally — reproducible companion

Figures 1–7 show documented official historical time series, not causal policy effects. Figure 8 is a hypothetical constant-inflation/indexation scenario, not observed wage data.


```python
from pathlib import Path
import numpy as np
from analysis import (FIGURE_NAMES, annual_average_inflation, build_all, fiscal_data,
    gini_data, illustrative_real_wage, macro_data, racial_unemployment_gap, union_data)
macro = macro_data()
inflation = annual_average_inflation(macro['cpi_u_annual_average_index'])
print('1980 annual-average CPI-U inflation:', round(inflation[10], 2))
print('1982 annual mean unemployment:', macro['unemployment_annual_average_percent'][13])
print('Pre-1972 Black annual means unavailable:', np.isnan(macro['black_unemployment_annual_average_percent'][:3]).all())
```

```python
gap = racial_unemployment_gap(macro['black_unemployment_annual_average_percent'], macro['white_unemployment_annual_average_percent'])
print('1983 Black–White unemployment gap, percentage points:', gap[14])
print('1980 and 1990 household Gini:', gini_data()['household_money_income_gini'][[13, -1]])
print('1983 and 1990 union rates:', union_data()['union_membership_rate_percent'][[0, -1]])
print('FY 1983 budget balance (% of fiscal GDP):', fiscal_data()['unified_surplus_deficit_percent_of_fiscal_gdp'][13])
```

```python
years = np.arange(0, 21)
print('Year-20 real wage index, no indexing:', illustrative_real_wage(years)[-1])
print('Year-20 real wage index, full indexing:', illustrative_real_wage(years, inflation_indexation=1)[-1])
```

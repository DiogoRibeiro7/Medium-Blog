# The Right to Vote Was Only the Beginning — reproducible companion

Figures 1–4 are published descriptive historical series, figure 5 is a sourced selected chronology, and figures 6–8 use wholly hypothetical inputs. No comparison here estimates the causal impact of legislation.


```python
from pathlib import Path
import numpy as np
from analysis import (FIGURE_NAMES, age_poverty, benefit_schedule, build_all,
    historical_events, medicare_enrollment, national_poverty, poverty_gap,
    racial_poverty, realized_voter_access)
national = national_poverty()
race = racial_poverty()
age = age_poverty()
print('Official 1959 vs 1975 all-person poverty rates:', national['poverty_rate_percent'][[0, -1]])
print('Black 1960–65 source entries missing:', np.isnan(race['black_poverty_rate_percent'][1:7]).all())
print('Older-age 1960–65 source entries missing:', np.isnan(age['age65plus_poverty_rate_percent'][1:7]).all())
print('1959 and 1975 racial gap (percentage points):', poverty_gap(race['white_poverty_rate_percent'], race['black_poverty_rate_percent'])[[0, -1]])
```

```python
cms = medicare_enrollment()
print('1966 hospital-insurance enrollees:', int(cms['total_hospital_insurance_enrollees'][0]))
print('1975 hospital-insurance enrollees:', int(cms['total_hospital_insurance_enrollees'][-1]))
print('Selected institutional events:', len(historical_events()))
```

```python
print('Invented registration access, barrier 40%:', realized_voter_access(100000, .85, .4))
earnings = np.array([18000., 18001.])
print('Illustrative hard-cliff resources:', earnings + benefit_schedule(earnings, cliff=True))
print('Illustrative tapered resources:', earnings + benefit_schedule(earnings, cliff=False))
```

```python
outputs = build_all(Path('.').resolve())
assert len(outputs) == len(FIGURE_NAMES) == 8
assert all(path.is_file() for path in outputs)
print('Eight figures and three machine-readable result tables reproduced.')
```

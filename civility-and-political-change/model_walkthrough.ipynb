# When Power Calls for Civility — a model walkthrough

This is a deterministic thought experiment about a weighted Nash division of **transferable** surplus. The quantities are invented to illustrate algebraic consequences. They are **not** political, historical, or social-science estimates. Read the [article](article.md) and the [reproducibility notes](README.md) before interpreting the charts.

The notebook imports the same tested functions as the figure-generating script. No external dataset or network access is used.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

from analysis import bargaining_solution, discounted_delay_loss, effective_outside_option, generate

print('Loaded deterministic mathematical model.')
```

## 1. Equal weights are not equal final outcomes

With total transferable surplus 1 and disagreement payoffs d_R and d_I, the reformer allocation is r = d_R + alpha*(1-d_R-d_I). We keep alpha=0.5 but vary the disagreement payoffs.

```python
examples: list[tuple[str, float, float]] = [
    ('Incumbent advantage', 0.05, 0.55),
    ('Balanced outside options', 0.30, 0.30),
    ('Reformer advantage', 0.55, 0.05),
]
for name, outside_r, outside_i in examples:
    reformer, incumbent = bargaining_solution(outside_r, outside_i, 0.5)
    print(f'{name:26} reformer={reformer:.2f} incumbent={incumbent:.2f}')
```

## 2. A conditional restraint-cost example

This section **assumes** that a procedural constraint reduces the reformer's disagreement payoff. Reversing that assumption can reverse the result; the notebook cannot establish whether the mechanism occurs in the world.

```python
for cost in (0.0, 0.5, 1.0):
    disagreement = effective_outside_option(0.35, cost, 0.30)
    reformer, _ = bargaining_solution(disagreement, 0.45, 0.5)
    print(f'assumed restraint={cost:.1f}: outside={disagreement:.3f}, negotiated={reformer:.3f}')
```

## 3. Waiting is not costless

Assuming one normalized unit of avoidable harm per year, its discounted value over T years at rate rho is (1-exp(-rho*T))/rho, with the zero-rate limit T.

```python
for years in (0, 5, 10, 20):
    value = discounted_delay_loss(float(years), 1.0, 0.03)
    print(f'{years:2d} years of delay: hypothetical present-value harm {value:.3f}')
```

## Interpretive boundary

The examples are transparent implications of a specified Nash bargaining model. They do **not** identify a causal effect of civility, determine whether an historical movement succeeded, or establish an ethically correct policy outcome. Those are separate empirical and normative tasks.
